```python
# Imports 

from assistant import ClayBankAssistant
from evaluator import ExpectedAnswerEvaluator
```

```python
# Define questions and expected answers

questions = [

    "Does the Clay Digital Account charge a monthly fee?",

    "Can customers receive salary payments in the Clay Digital Account?",

    "What is the nighttime instant transfer limit?",

    "What are possible reasons for a declined credit card purchase?",

    "How long does it take to receive loan funds after approval?",

    "Can customers pay off the personal loan early?",

    "Is a suitability assessment mandatory before investing?",

    "What documents are required to open an insurance claim?",

    "Can support agents request the customer's full password?",

    "What should be done in case of fraud suspicion?"
]

expected_answers = [

    "No. The Clay Digital Account is free of charge.",

    "Yes. Salary portability is supported.",

    "The nighttime instant transfer limit is $1,000 per day.",

    "Possible reasons include insufficient limit, preventive block, fraud suspicion, or expired card.",

    "Funds are disbursed up to 2 hours after approval.",

    "Yes. Early repayment is allowed with proportional discount on future interest.",

    "Yes. A suitability assessment is mandatory before the first investment.",

    "Required documents include police report, personal identification, and event supporting documents.",

    "No. Support agents must never request the customer's full password.",

    "The process includes blocking the card, ending active sessions, validating identity, and opening an anti-fraud protocol."
]
```

```python
# Runing the assistant

assistant = ClayBankAssistant()

agent_answers = []

for question in questions:

    answer = assistant.ask(question)

    agent_answers.append(answer)

    print(f"QUESTION:\n{question}\n")
    print(f"ANSWER:\n{answer}\n")
    print("--------------------------------------------------\n")
```

```python
# Running the evaluator

evaluator = ExpectedAnswerEvaluator()

scores = []

for i in range(len(questions)):

    score = evaluator.evaluate(
        question=questions[i],
        agent_response=agent_answers[i],
        expected_response=expected_answers[i]
    )

    scores.append(score)

    print(f"QUESTION {i+1}")
    print(f"SCORE: {score}")
    print("--------------------------------------------------")
```

```python
# Final score calculation

final_score = sum(scores) / len(scores)

print("\n==============================")
print("FINAL QUALITY SCORE")
print("==============================\n")

print(f"Score: {final_score:.2f}")

print(f"Correct Answers: {sum(scores)}")
print(f"Total Questions: {len(scores)}")
```
